```python
#Análise de dupla massa entre a Estação Joinville e Itapocu
```

```python
# Selecionando apenas as colunas de Data e total de precipitação
dataset1 = dataset1[['Data', 'Total']]
dataset2 = dataset2[['Data', 'Total']]
```

```python
# Convertendo a coluna de data
dataset1['Data'] = pd.to_datetime(dataset1['Data'], dayfirst=True, errors='coerce')
dataset2['Data'] = pd.to_datetime(dataset2['Data'], dayfirst=True, errors='coerce')
```

```python
# Unindo as duas séries pela data
# ---------------------------------------------------------
df = pd.merge(dataset1, dataset2, on='Data', suffixes=('_Joinville', '_Itapocu'))
```

```python
df
```

```python
# Eliminando linhas com valores ausentes
df = df.dropna()
df
```

```python
# Calculando as chuvas acumuladas (para fazer a curva de dupla massa)
df['Acumulada_Joinville'] = df['Total_Joinville'].cumsum()
df['Acumulada_Itapocu'] = df['Total_Itapocu'].cumsum()
```

```python
# Cálculo da regressão linear para verificar a consistência
```

```python
coef = np.polyfit(df['Acumulada_Joinville'], df['Acumulada_Itapocu'],1)
reta = np.poly1d(coef)

print(f"Inclinação (coeficiente angular): {coef[0]:.3f}")


plt.figure(figsize=(8,8))
plt.scatter(df['Acumulada_Joinville'],df['Acumulada_Itapocu'], color='blue', label='Dados observados')
plt.plot(df['Acumulada_Itapocu'], reta(df['Acumulada_Itapocu']), 'r--', label='Tendência linear')
plt.ylabel('Precipitação Acumulada - Itapocu (mm)')
plt.xlabel('Precipitação Acumulada - Joinville (mm)')
plt.title('Curva de Dupla Massa com Regressão Linear')
plt.legend()
plt.grid(True)
plt.show()
```
